```python
metadata_file = "combined_UCE_5neuro/obs_anno.tsv"
data = pd.read_csv(metadata_file, sep ='\t')
data.columns
```

```python
#composite new column

# Define your conditions
conditions = [
    data["publication_label"] == "Gabitto et al. (2024) Nat Neurosci",
    data["publication_label"] == "Siletti et al. (2023) Science",
    data["publication_label"] == "BICAN Basal Ganglia",
]

# Define the corresponding column to pull from
choices = [
    data["cell_type_ontology_term_id"],
    data["cell_type_ontology_term_id"],
    data["CL:ID_group"],
]

# Create the new column
data["CL"] = np.select(
    conditions, choices, default=""
)

# Keep only values starting with "CL:"; everything else -> ""
data["CL"] = data["CL"].where(data["CL"].astype(str).str.startswith("CL:"), "")
```

```python
unique_rows = data[["CL", "Gabitto_cell_type/Siletti_supercluster/BG_group"]].drop_duplicates()
print(unique_rows.to_string(index=False))
```

```python
mapping = {
    "Mammillary body": "CL:4023074",
    "Thalamic excitatory": "CL:4023068",
    "Amygdala excitatory": "CL:4023039",
    "Medium spiny neuron": "CL:1001474",
    "Eccentric medium spiny neuron": "CL:4030057",
    "Cerebellar inhibitory": "CL:4023057",
    "Midbrain-derived inhibitory":"CL:4023079",
    "CGE interneuron":"CL:4023064",
    "MGE interneuron":"CL:4023063",
    "Deep-layer near-projecting":"CL:4030067",
    "Deep-layer corticothalamic and 6b":"CL:4023042",
    "Hippocampal CA1-3":"CL:4023060",
    "Upper-layer intratelencephalic":"CL:4023008",
    "Deep-layer intratelencephalic":"CL:4023008",
    "Hippocampal dentate gyrus":"CL:4023062",
    "Hippocampal CA4":"CL:4023061",
}
data["CL"] = data["supercluster_term"].map(mapping).fillna(data["CL"])
```
