## latest_census_datasets

```python
latest_census_datasets = latest_census["census_info"]["datasets"].read().concat().to_pandas()
```

## get dataset ids and info from our current db

```python
already_have_dataset_list =[]

if embedding=="UCE":
    local_dir = "UCE_npy_obs"
    suffix = "_uce.npy"
elif embedding=="scimilarity":
    local_dir = "SCimilarity_npy_obs"
    suffix = "_scimilarity.npy"
for root, dirs, files in os.walk(local_dir):
    for file in files:
        if file.startswith("ucsc-"):
            continue
        if file.endswith(suffix):
            dataset_id = file.removesuffix("_uce.npy")
            match = re.match(r'^([a-f0-9\-]+)_', file)
            real_dataset_id = match.group(1)
            if real_dataset_id not in already_have_dataset_list:
                already_have_dataset_list.append(real_dataset_id)
len(already_have_dataset_list)
```

```python
df = latest_census_datasets[latest_census_datasets["dataset_id"].isin(already_have_dataset_list)]
df = df[["dataset_id", "dataset_total_cell_count", "collection_name", "dataset_title", "collection_doi_label","collection_doi"]].sort_values(by="dataset_total_cell_count", ascending=True)
df
```

```python
import datetime
today = datetime.date.today()

output = f"datasets.info.cxg.{embedding}.{today}.txt"
with open(output, "w") as f:
    df.to_csv(f, sep="\t", index=False)
```
