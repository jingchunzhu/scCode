### Python code to call the OpenAI API developmental_stage

```python
counts = meta["development_stage"].value_counts()
counts
```

```python
# Step 1: group and count
df_stage_counts = (
    meta
    .groupby(
        ["development_stage_ontology_term_id", "development_stage"],
        dropna=False
    )
    .size()
    .reset_index(name="count")  # <-- convert Series to DataFrame here
)
df_stage_counts
```

```python
development_stage_entries = df_stage_counts.to_dict(orient="records")

user_message = {
    "development_stage_entries": development_stage_entries
}
```

```python
system_prompt = """
You are a biological metadata normalization assistant.

Your task is to normalize heterogeneous "development_stage" entries into standardized age buckets.

You will be given both:
- the ontology term ID ("development_stage_ontology_term_id")
- the raw label ("development_stage")

Instructions:
- Use the ontology term ID as the primary source of truth whenever it is available.
- Use the raw label only if the ontology ID is missing, unknown, or cannot be interpreted.
- Normalize each entry into a standardized age bucket as defined below.
- Do not make assumptions beyond the information provided.

Your input may include:
- exact ages (e.g. "42-year-old human stage")
- age ranges (e.g. "25–44 year-old human stage")
- decades (e.g. "sixth decade stage")
- developmental terms (e.g. "embryonic human stage", "adult stage")
- prenatal timing (e.g. "16th week post-fertilization", "Carnegie stage 20")
- infancy or childhood descriptions
- unknown or ambiguous labels

---

### STANDARDIZED AGE BUCKETS (ordered):

- prenatal_embryonic            (blastula, Carnegie stages, weeks post-fertilization ≤ 8)
- prenatal_fetal                (weeks post-fertilization > 8 until birth)
- newborn                        (0–28 days)
- infant                         (1 month – <1 year)
- child                          (1–12 years)
- adolescent                     (13–17 years)
- young_adult                    (18–39 years)
- middle_adult                   (40–59 years)
- older_adult                    (60–79 years)
- elderly                        (80+ years)
- adult_unspecified              (labels like "adult", "human adult stage")
- postnatal_unspecified          (postnatal but no age resolution)
- unknown                        (explicitly unknown or cannot be inferred)

---

### Classification rules:

Step 0: Unknown (highest priority)
- If the ontology ID or label indicates "unknown", "unclassified", or cannot be interpreted → assign "unknown".

Step 1: Prenatal stages
- Carnegie stages → prenatal_embryonic
- Blastula, embryonic → prenatal_embryonic
- Post-fertilization weeks:
    - ≤ 8 weeks → prenatal_embryonic
    - > 8 weeks → prenatal_fetal
- LMP month references → prenatal_fetal

Step 2: Newborn / infant
- "newborn", "0–28 days" → newborn
- months old (<12 months) → infant
- "under-1-year-old" → infant

Step 3: Childhood / adolescence
- 1–12 years → child
- 13–17 years → adolescent

Step 4: Adults (numeric or decade-based)
- 18–39 → young_adult
- 40–59 → middle_adult
- 60–79 → older_adult
- ≥ 80 → elderly
- Decade-based terms (e.g. "sixth decade") should be mapped accordingly.

Step 5: Adult but unspecified
- Labels like "adult stage", "human adult stage", "early adulthood", "late adulthood"
  → adult_unspecified

Step 6: Postnatal but unspecified
- "postnatal stage", "infant stage" (without age)
  → postnatal_unspecified

---

### Output format (JSON ONLY):

Return a JSON object with this structure:

{
  "annotations": [
    {
      "development_stage_ontology_term_id": "...",
      "original_label": "...",
      "normalized_age_bucket": "..."
    }
  ]
}

Each input entry must appear exactly once.  
Do not include explanations or commentary.


"""
```

```python
client = OpenAI()

response = client.responses.create(
    model="gpt-4.1", # "gpt-4o"
    input=[
        {"role": "system", "content": system_prompt},
        {"role": "user", "content": json.dumps(user_message)}
    ]
)

# Extract JSON from the response
content = response.output_text
annotations = json.loads(content)["annotations"]

# Convert back to DataFrame
df_annotations = pd.DataFrame(annotations)

df_annotations
```

```python
df_annotations_with_counts = df_annotations.merge(
    df_stage_counts[['development_stage_ontology_term_id', 'development_stage', 'count']],
    left_on=['development_stage_ontology_term_id', 'original_label'],
    right_on=['development_stage_ontology_term_id', 'development_stage'],
    how='left'
)
df_annotations_with_counts = df_annotations_with_counts.drop(columns=['original_label', 'development_stage_ontology_term_id'])

age_bucket_order = [
    "prenatal_embryonic",
    "prenatal_fetal",
    "newborn",
    "infant",
    "child",
    "adolescent",
    "young_adult",
    "middle_adult",
    "older_adult",
    "elderly",
    "adult_unspecified",
    "postnatal_unspecified",
    "unknown"
]
# Convert 'normalized_age_bucket' to a categorical type with this order
df_annotations_with_counts['normalized_age_bucket'] = pd.Categorical(
    df_annotations_with_counts['normalized_age_bucket'],
    categories=age_bucket_order,
    ordered=True
)

# Sort the DataFrame by this custom order
df_annotations_with_counts_sorted = df_annotations_with_counts.sort_values('normalized_age_bucket').reset_index(drop=True)

df_annotations_with_counts_sorted
```

```python
bucket_cell_counts = (
    df_annotations_with_counts_sorted
    .groupby('normalized_age_bucket', observed=True)['count']
    .sum()
    .reset_index()
)

bucket_cell_counts
```

```python
len(bucket_cell_counts['count'])
```

```python
# Create a figure larger than the axes to accommodate margins
fig = plt.figure(figsize=(6, 3))  # Figure slightly bigger than axes

# Add axes with exact size in inches
# [left, bottom, width, height] in figure coordinates (0-1)
# We can compute figure coords to get axes 5x2 inches
fig_width, fig_height = fig.get_size_inches()
ax_width, ax_height = 5, 2
left = 0.1  # adjust as needed
bottom = 0.1  # adjust as needed

ax = fig.add_axes([
    left, 
    bottom, 
    ax_width / fig_width, 
    ax_height / fig_height
])

# Plot
ax.bar(
    bucket_cell_counts['normalized_age_bucket'],
    bucket_cell_counts['count']
)

ax.set_xlabel("")
ax.set_ylabel("Cell count")
ax.set_title("normalized_age_bucket")

plt.xticks(rotation=90, ha="right")
plt.show()
```

### annotate metadata with normalized_age_bucket

```python
# check 1-to-1 mapping is good, expect no >1 values
(df_annotations_with_counts_sorted
 .groupby('development_stage')['normalized_age_bucket']
 .nunique()
 .sort_values(ascending=False))
```

```python
age_map = (
    df_annotations_with_counts_sorted[
        ['development_stage', 'normalized_age_bucket']
    ]
    .drop_duplicates(subset='development_stage')
)
age_map
```
