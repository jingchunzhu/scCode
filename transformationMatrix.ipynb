```python
import cv2
import numpy as np
import pandas as pd
from skimage.io import imread
```

```python
image_path = 'download_usefuldata/Xenium_V1_hPancreas_Cancer_Add_on_FFPE_if_image.ome.tif'
transformation_matrix_path = 'download_usefuldata/Xenium_V1_hPancreas_Cancer_Add_on_FFPE_if_imagealignment.csv'
output_path = "IF_image.tif"
target_image_path = "download_usefuldata/morphology_mip.ome.tif"
```

```python
original_image = imread(image_path)
```

```python
target_image = imread(target_image_path)
```

```python
h, w = target_image.shape
```

# Transformation with kwimage

```python
transformation_matrix = pd.read_csv(transformation_matrix_path, header=None).values
transformation_matrix
```

```python
from kwimage.im_cv2 import *  # NOQA
import kwimage
from kwimage.transform import Affine
```

```python
# https://kwimage.readthedocs.io/en/release/kwimage.html#kwimage.warp_affine

# large_warp_dim (int | None | str) – If specified, perform the warp piecewise in chunks of the specified size. 
# If “auto”, it is set to the maximum “short” value in numpy. This works around a limitation of cv2.warpAffine, which must have image dimensions < SHRT_MAX (=32767 in version 4.5.3)

warped1 = warp_affine(original_image, transformation_matrix, dsize= (w, h), large_warp_dim='auto')
```

```python
cv2.imwrite(output_path, warped1)
```
