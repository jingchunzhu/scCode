```python
import numpy
from skimage.io import imread
import cv2, json, pandas as pd
```

```python
image_path = '/mnt/windows/16-080L/CosMx/TMA_100C_CosMxRNA_HE_v2.tif'
output_path = 'HE_16-080L.tif'
geojsonfile = "HE_16-080L.geojson"
```

```python
original_image = imread(image_path)
```

```python
original_image.shape
```

## Read geojsonfile (from Qupath single rectangle annotation)


```python
J = json.load(open(geojsonfile, 'r'))
rectangle = pd.DataFrame(J["features"][0]["geometry"]["coordinates"][0])
# x
x_min = min(rectangle[0]) 
x_max = max(rectangle[0]) 
# y
y_min = min(rectangle[1]) 
y_max = max(rectangle[1]) 
```

```python
x_min, x_max, y_min, y_max
```

## make small images

```python
small_array = original_image[y_min:y_max, x_min:x_max]
```

```python
small_array.shape
```

```python
import cv2
```

```python
# Flipping the image around x-axis - this particular image needs flip top and bottom
# https://datahacker.rs/003-how-to-resize-translate-flip-and-rotate-an-image-with-opencv/
# 1 for vertical LEFT_RIGHT
# 0 for horizontal flip TOP_BOTTOM

flipped = cv2.flip(small_array, 0)
```

```python
flipped.shape
```

```python
cv2.imwrite(output_path, flipped)
```
